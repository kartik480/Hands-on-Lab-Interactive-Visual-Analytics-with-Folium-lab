```python
# Step 1: Import required libraries
import folium
import pandas as pd
import io
from js import fetch
from folium.features import DivIcon

# Step 2: Download the dataset
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'

resp = await fetch(URL)
spacex_csv_file = io.BytesIO((await resp.arrayBuffer()).to_py())

spacex_df = pd.read_csv(spacex_csv_file)

# Step 3: Select required columns
spacex_df = spacex_df[
    ['Launch Site', 'Lat', 'Long', 'class']
]

# Step 4: Create launch_sites_df
launch_sites_df = spacex_df.groupby(
    ['Launch Site'], as_index=False
).first()

launch_sites_df = launch_sites_df[
    ['Launch Site', 'Lat', 'Long']
]

# Step 5: Define NASA coordinates
nasa_coordinate = [29.559684888503615, -95.0830971930759]

# Step 6: Initialize map
site_map = folium.Map(
    location=nasa_coordinate,
    zoom_start=5
)

# Step 7: Add all launch sites
for index, row in launch_sites_df.iterrows():

    coordinate = [row['Lat'], row['Long']]

    circle = folium.Circle(
        location=coordinate,
        radius=1000,
        color='#000000',
        fill=True
    ).add_child(
        folium.Popup(row['Launch Site'])
    )

    marker = folium.map.Marker(
        coordinate,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html='<div style="font-size:12px;color:#d35400;"><b>%s</b></div>'
            % row['Launch Site']
        )
    )

    site_map.add_child(circle)
    site_map.add_child(marker)

# Step 8: Display map
site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):

    # Approximate radius of Earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c

    return distance
```
